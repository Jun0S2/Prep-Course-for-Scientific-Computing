## Compressed row storage (CRS) for sparse matrices

Define CRS data structure for 

$$
A=\begin{bmatrix} 0 & 0 & 3 \\ 4 & 0 & 0 \\ 0 & 5 & 6 \end{bmatrix}:
$$

```python
values = [3, 4, 5, 6]
col_index = [2, 0, 1, 2]
row_ptr = [0, 1, 2, 4]
```

Define function for sparse matrix-vector multiplication:

```python
def vmult(dst, src):
    for row in range(len(row_ptr)-1):
        result = 0.0
        for col in range(row_ptr[row], row_ptr[row+1]):
            result += values[col] * src[col_index[col]]
        dst[row] = result
```

Perform sparse matrix-vector multiplication:

```python
src = [1, 1, 1]
dst = [0, 0, 0]

vmult(dst, src)

print(dst)
```
