## In-class task 10.1: Newton's method

In this task, we solve the nonlinear problem

$$F(x, y) = \begin{bmatrix}
x^2 + y^2 - 4 \\
e^x+ y - 1
\end{bmatrix} = \begin{bmatrix}
0 \\ 0
\end{bmatrix},$$

using Newton's method.

### Introduction


We start with importing the relevant packages.

```python
import math
import numpy as np
```

### Algorithm

Implement the Newton's method in the next box:

```python
def newton(fu_rhs, x, fu_jac, n_max_iterations = 100, tolerance = 1e-10):
    """
    newton Newton solver.

    :param fu_rhs: function for which the root should be found
    :param x: start guess
    :param fu_jac: function that returns the Jacobian for a given linerization point
    :param n_max_iterations: maximal number of iterations
    :param tolerance: l2 norm of residual to be reached
    :return: solution after one time step
    """ 
    # create array for x
    x = np.array(x, dtype=float)
    
    for i in range(n_max_iterations):
        F = fu_rhs(x) # call fu_rhs(x)  returns val at x 
        J = fu_jac(x) # call fu_jac(x)  returns val at x
        
        # Solve J * delta = -F
        delta = np.linalg.solve(J, -F) # Jδ=−F https://numpy.org/devdocs/reference/generated/numpy.linalg.solve.html
        x = x + delta # next guess
        
        # Check convergence
        # residual tells you how off you are from the guess
        # if it converges to 0
        #ord - order
        if np.linalg.norm(F, ord=2) < tolerance:
            print(f"Converged in {i+1} iterations.")
            return x

    raise ValueError("Newton method did not converge within the maximum number of iterations.")
```

### Application

Now, let's apply the three implemented functions to solve the given nonlinear problem. Fill in the following boxes containing the residual function:

```python
def fu_rhs(sol):
    x, y = sol
    return np.array([
        x**2 + y**2 - 4, # first equation
        np.exp(x) + y - 1 # second equation
    ])
```

a function that returns its Jacobian:

take derivates

```python
def fu_jac(sol):
    x, y = sol
    return np.array([
        [2*x, 2*y],
        [np.exp(x), 1]
    ])
```

and the initial guess $x_0=1$, $y_0=1$:

```python
start = np.array([1.0, 1.0])
```

Now, let's run the code:

```python
result = newton(fu_rhs, start, fu_jac)

print("The result is: %f, %f" % (result[0], result[1]))
```
