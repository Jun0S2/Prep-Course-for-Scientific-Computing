### Vectorization

Run the following code snippet. It performs a simple vector operation with a for-loop on a Python list (_non-vectorized code_) and with a NumPy array (_vectorized code_) and measures the running times. Try different sizes of the vector. Pay attention to differences in runtime and clarity of the code (i.e., how easy the code is to write and understand).

```python
# Vector size
N = int(1e7)

# Python list
startTime_list = time.time()

x_list = list(range(N))
y_list = [math.sin(2*math.pi/N*x) for x in x_list]

endTime_list = time.time()

# Numpy array
startTime_arr = time.time()

x_arr = np.arange(0,N)
y_arr = np.sin(2*math.pi/N*x_arr)

endTime_arr = time.time()

print("Elapsed time with list: {:f} seconds\nElapsed time with NumPy array: {:f} seconds".format(endTime_list-startTime_list,endTime_arr-startTime_arr))
```

Note: _vectorization_ is not only referring to replacing for-loops by vector operations but also to rewriting your algorithm so that you can use the right dense linear-algebra routines referred to as BLAS (Basic Linear Algebra Subprograms) routines. These can be categorized as follows:
-  Level 1 BLAS: vector operations, e.g., $$y \gets \alpha x + y,$$
-  Level 2 BLAS: matrix-vector operations, e.g., $$y \gets \alpha A x + \beta y,$$
-  Level 3 BLAS: matrix-matrix operations, e.g., $$C \gets \alpha AB + \beta C.$$

These BLAS routines are highly optimized for most hardware architectures (e.g., SIMD vectorization, cache blocking) and as a result fast. The key message is: use built-in functions if possible.

### Testing

Throughout the course, you will need to implement different functions. We will help you by providing **tests**. Tests compare, for a given input, the output to an expected output.

Below, an example is given where a function `fu` is defined and tested.


```python
def fu(values):
    return values * 2

input = np.array([0, 1, 2, 3])
output = np.array([0, 2, 4, 6])

assert np.all(np.isclose(fu(input), output))
```
