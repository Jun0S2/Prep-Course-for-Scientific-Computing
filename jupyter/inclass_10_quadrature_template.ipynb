```python
def midpoint(func, a, b, n):
    """
    midpoint applies the composite midpoint rule for integrating a function for specified bounts.

    :param func: the function that should be integrated
    :param a: start point of integration interval
    :param b: end point of integration interval
    :return: n: number of subintervals
    """ 
    h = (b-a)/n
    x = np.linspace(a, b, n+1)
    x = x[:-1] + h/2
    fx = func(x)
    M = np.sum(fx)
    return h * M
```

Now, let's define the function we want to integrate:

```python
def humps(x):
    return 4 / (x**2 + 1)
```

The function above can be integrated analytically.

```python
def humps_int(a, b):
    return 4 * (np.arctan(b) - np.arctan(a))
```
