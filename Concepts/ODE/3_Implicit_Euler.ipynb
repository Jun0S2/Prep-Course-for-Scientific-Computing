# Implicit Euler
> 한걸음 앞의 기울기를 이용해야해서, 방정식을 풀어야함.

$$
y_{n+1} = y_{n} = h*f(t_{n+1},y_{n+1})
$$

멍청이슈
$$
 \frac{dy}{dt} = f(t,y) = -2y
$$

또, h 는 시간 (step) 이고, $f(t_{n},y_{n})은 속도(기울기) 임.
즉, `시간 x 속도 = 거리`, 즉, `y 의 변화량`이 되는거임

### Step 0
> t = 0, y = 1

$$
f(t,y) = -2y
$$

0.1초 동안 기울기 -2 만큼 이동 -> y 의 변화량 : 0.1x(-2) = -0.2

* $ y_{1} = y_{0} + h * f(t_{1}, y_{1}) $
* $ y_{1} = 1 + 0.1 * -2 * y_{1}  = 0.83 $

## Step 1
> t = 0.1, y1 = 0.83

$$
y_{2}   = y_{1} + h(t2, y2)
$$

$$
        = 0.8 + 0.1 (-2 * 0.83)
$$

$$
        = 0.694
$$

Repeat steps , we get y(0.4) === 0.4823

### Conclusion
Implicit Euler $ y(0.4) == 0.4823$
