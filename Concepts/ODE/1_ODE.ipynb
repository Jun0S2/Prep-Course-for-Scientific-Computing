# ODE (Ordinary Differential Equation)

미분 방정식 $ y'(t) = f(t,y) $ 를 수치적으로 푸는 것
> 실제 해 (analytic Solution)을 구할 수 없는 경우가 많은데, 이때 근사값을 구하는 numerical methods를 사용하는 것임!

---

## Numerical Methods

그렇다면, 어떤 numerical methods 가 있을까?

| Methods  | types  |  descriptions | stability/convergence  |
|---|---|---|---|
| Explicit Euler  | explicity  | $ y_{n+1} = y_{n} + h*f(t_{n}, y_{n})  $ | simple, unstable , order1 |
|  Implicit Euler | implicity  | $ y_{n+1} = y_{n} + h*f(t_{n+1}, y_{n+1})  $   | 비선형방정식으로 풀어야 함  |
|  Heun's Method(improved Euler) | explicity  |평균 기울기 사용. $ y_{n+1} = y_{n} + \frac{h}{2}*(f_{n} + f^{*}_{n+1})  $   |  stability increase, order 2 |
|  Runge-Kutta 4th Order | explicit  | 4개의 중간 기울기 평균  |  정확도 4차, order 4 |

explicit : 다음 값을 바로 계산, implicit : 다음 값이 식에 포함됨


### 용어 정리

1. stability : 작은 오차가 커지지 않는다
2. convergence : global error -> 0, as h-> 0
3. consistency : local error -> 0, as h->0

ODE에서의 한 step 의 오차가, 다음 step에 영향을 준다.

### 예제 문제
$$
y^{'}(t) = f(t, y(t)) , y(t_{0}) = y_{0}
$$

* step size : h
* 목표 : y_{n+1} 근사값 계산

1. 변수분리
$$
\frac{dy}{dt} = -2y
$$

$$
\frac{1}{y} \, dy = -2 \, dt
$$

2. 양변 적분
$$
\int \frac{1}{y} \, dy = \int -2 \, dt
$$

3. 적분 계산
$$
\ln|y| = -2t + C
$$

4. 지수함수로 풀기
$$
y = e^{-2t + C} = e^C \cdot e^{-2t}
$$

5. 초기조건 적용 (t=0일 때 y=1)
$$
1 = e^C \cdot e^{0} = e^C
$$

6. 최종 해
$$
y = e^{-2t}
$$

이제, Euler (explicit, implicit), Heun's, RK4로 같은 문제를 풀어보겠다!
