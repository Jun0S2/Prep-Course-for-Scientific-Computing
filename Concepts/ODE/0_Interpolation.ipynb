# Interpolation

주어진 점들  (data) 사이를 이을 수 있는 함수를 찾는 법
-> error estimation에서 자주 사용된다.

- Interpolation error 공식을 사용해서 오차의 크기를 분석
- 적분/미분/ODE 등 사용한다.

> EX) Taylor Series, interpolation series

#### General Interpolation Error
$$ f(x) - P_n(x) = \frac{f^{(n+1)}(\xi)}{(n+1)!} \prod_{i=0}^{n} (x - x_i) $$

#### Linear Interpolation Error
$$ |f(x) - P_1(x)| \leq \frac{(x_1 - x_0)^2}{8} \max_{\xi \in [x_0, x_1]} |f''(\xi)| $$

#### 등간격 노드에서의 오차 한계

$$
|f(x) - P_n(x)| \leq \frac{h^{n+1}}{4(n+1)} \max_{\xi \in [a,b]} |f^{(n+1)}(\xi)|
$$
