# Explicit Euler
> 현재 기울기(위치) -> step ++

$$
y_{n+1} = y_{n} = h*f(t_{n},y_{n})
$$

멍청이슈
$$
 \frac{dy}{dt} = f(t,y) = -2y
$$

또, h 는 시간 (step) 이고, $f(t_{n},y_{n})은 속도(기울기) 임.
즉, `시간 x 속도 = 거리`, 즉, `y 의 변화량`이 되는거임

### Step 0
> t = 0, y = 1

$$
f(t,y) = -2y
$$

0.1초 동안 기울기 -2 만큼 이동 -> y 의 변화량 : 0.1x(-2) = -0.2

* $ y_{1} = 1 + (-0.2) = 0.8 $
* $ y_{1} = y_{0} + h * f(t_{0}, y_{0}) $

## Step 1
> t = 0.1, y1 = 0.8

$$
y_{2}   = y_{1} + h(t1, y1)
$$

$$
        = 0.8 + 0.1 (-2 * 0.8)
$$

$$
        = 0.64 
$$

## Step 2
> t2 = 0.2, y2 = 0.64

$$
y_{3} = y_{2} + h*f(t_{2}, y_{2}) 
$$

$$
= 0.64+0.1 * (-2*0.64) = 0.512
$$

## Step 3
> t2 = 0.3, y2 = 0.512

$$
y_{4} = y_{3} + h*f(t_{3}, y_{3}) 
$$

$$
= 0.512+0.1 * (-2*0.512) = 0.4096
$$

### Conclusion
Explicity Euler $ y(0.4) == 0.496$
