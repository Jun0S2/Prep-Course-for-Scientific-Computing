# Heun's Method


### 공식
#### 1. Prediction
$$
y_{n+1} = y_{n} = h*f(t_{n},y_{n})
$$

#### 2. Corrector
$$
y_{n+1} = y_{n} = \frac{h}{2}*f(t_{n},y_{n} + f(t_{n+1}, y_{n+1}))
$$


---
### Step 0
> t0 = 0.1, y0 = 1

1. Predictor
$$
y_{1} = 1 + 0 * -2 = 0.8
$$

2. Corrector
$$
y_{2} = 1 + \frac{0.1}{2} * (-2*-0.8) = 0.82
$$

---
### Step 1
> t1 = 0.1, y1 = 0.82

1. Predictor
$$
y_{2} = 0.82 + 0.1 * (-2*0.82) = 0.656
$$

2. Corrector
$$
y_{2} = 83 + \frac{0.1}{2} * (-2*-0.82 + -2*0.656) = 0.674
$$

### Conclusion

Repeat the process, and we get y3 = 0.5514, y4 = 0.4521
