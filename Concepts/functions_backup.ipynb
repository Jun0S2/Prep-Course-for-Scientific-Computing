# 함수, 상(image), 역상(preimage) 정리 + 예제 풀이
---

## 1. 기본 개념
함수는 **입력(정의역)** → **출력(공역)** 으로 값을 대응시키는 규칙이다.

$$
f : X \to Y, \quad x \mapsto f(x)
$$

- **정의역 (domain)** : 입력 값들의 집합 $ ( X ) $
- **공역 (codomain)** : 가능한 출력 값들의 전체 집합 $ ( Y ) $  
- **상 (image)** : 실제로 나오는 값들의 집합 $( f(X) \subseteq Y ) $

---

## 2. 수학 기호 정리

| 기호 | 정의 | 의미 (한글) | 영어 표현 |
|------|-------|--------------|------------|
| $\mathbb{R}$ | 모든 실수 | 실수 전체 | real numbers |
| $\mathbb{R}_+$ | $\{x \in \mathbb{R} : x \ge 0\}$ | 0 이상인 실수 | nonnegative reals |
| $\mathbb{R}_-$ | $\{x \in \mathbb{R} : x \le 0\}$ | 0 이하인 실수 | nonpositive reals |
| $f(X)$ | $\{f(x) : x \in X\}$ | 상(image) | image of $X$ |
| $f^{-1}(A)$ | $\{x \in X : f(x) \in A\}$ | $A$의 **역상(preimage)** | preimage of $A$ |

---

> ⚠️ 주의:  
> 여기서 $f^{-1}(A)$ 는 “**역함수**(inverse function)”이 아니라  
> “**역상 집합**(preimage)”을 뜻한다!

---

## 3. 함수의 종류 (injective / surjective / bijective)

| 이름 | 한글 | 뜻 | 예시 |
|------|------|----|------|
| **injective** | 단사 함수 | 서로 다른 입력이 서로 다른 출력으로 간다. | $f(x)=x+1$ |
| **surjective** | 전사 함수 | 가능한 모든 출력(공역의 원소)을 실제로 만든다. | $f(x)=x^3$ |
| **bijective** | 전단사 함수 | 단사 + 전사 (즉, 1:1 대응) | $f(x)=x+1$ |

## 4. Task 12.1

**문제:**  
$$
g : \mathbb{R} \to \mathbb{R}, \quad 
g(x) =
\begin{cases}
0, & x \le 0,\\[4pt]
1, & x > 0
\end{cases}
$$

$$
\mathbb{R}_- := \{x \in \mathbb{R} : x \le 0\}
$$
일 때,  
$$ 
g(\mathbb{R}),\ g^{-1}(\mathbb{R}_-),\ g^{-1}(\{0\})
$$
를 구하여라.

---

### ✏️ 풀이

- $ g(\mathbb{R}) = \{0, 1\} $  
  (모든 실수를 넣으면 0 또는 1만 나옴)

- $ g^{-1}(\mathbb{R}_-) = \{x \in \mathbb{R} : g(x) \in \mathbb{R}_-\}$  
  → $g(x) \le 0$ 이려면 $g(x) = 0 \Rightarrow x \le 0$
  ⇒ **$ g^{-1}(\mathbb{R}_-) = \mathbb{R}_-\ $**

- $ g^{-1}(\{0\}) = \{x : g(x)=0\} = \{x \le 0\} = \mathbb{R}_-\ $

---

✅ **결과 요약**

| 항목 | 값 |
|------|----|
| $g(\mathbb{R})$ | $\{0,1\}$ |
| $g^{-1}(\mathbb{R}_-)$ | $\mathbb{R}_-$ |
| $g^{-1}(\{0\})$ | $\mathbb{R}_-$ |

## Task 12.2

$$
\mathbb{R}_+ := \{x \in \mathbb{R} : x \ge 0\}, \quad f(x)=x^2
$$

다음 각각에 대해 함수가 단사/전사/전단사인지 판별하라.

---
### 1️. $f:\mathbb{R}\to\mathbb{R}$

- 단사 ❌ ($f(-1)=f(1)$)  
- 전사 ❌ (음수값 불가능)  
→ **둘 다 아님**

---

### 2️. $f:\mathbb{R}\to\mathbb{R}_+$

- 단사 ❌ (±값 겹침)  
- 전사 ✅ (모든 $y \ge 0$에 대해 $\pm\sqrt{y}$ 존재)  
→ **전사만**

---

### 3️. $f:\mathbb{R}_+\to\mathbb{R}$

- 단사 ✅ ($x, y \ge 0$이면 $x^2 = y^2 \Rightarrow x = y$)  
- 전사 ❌ (음수는 불가능)  
→ **단사만**

---

### 4️. $f:\mathbb{R}_+\to\mathbb{R}_+$

- 단사 ✅  
- 전사 ✅ (모든 $y \ge 0$에 대해 $x = \sqrt{y} \in \mathbb{R}_+$)  
→ **전단사(bijective)**

---

✅ **요약표**

| 경우 | 정의 | 단사? | 전사? | 비고 |
|------|--------|--------|--------|--------|
| 1 | $f:\mathbb{R}\to\mathbb{R}$ | ❌ | ❌ | 둘 다 아님 |
| 2 | $f:\mathbb{R}\to\mathbb{R}_+$ | ❌ | ✅ | 전사만 |
| 3 | $f:\mathbb{R}_+\to\mathbb{R}$ | ✅ | ❌ | 단사만 |
| 4 | $f:\mathbb{R}_+\to\mathbb{R}_+$ | ✅ | ✅ | 전단사(bijective) |
